```python
# Consultas Analíticas feitas em memória
df.createOrReplaceTempView("vw_emendas_temp")

# Consulta para validar a granularidade dos codigos de emenda, existem duplicidades de emendas com o mesmo código,
# Porém com os demais valores diferentes, não se trata de um erro, pode ser que o mesmo código de emenda seja utilizado
# Para diferentes objetivos.

# Exemplos: Fases Diferentes da Despesa (Empenho, Liquidação e Pagamento), Descentralização ou Múltiplos Beneficiários ou Modificações e Remanejamentos Orçamentários
spark.sql("""
    SELECT 
        `Código_da_Emenda`, 
        COUNT(*) AS total_ocorrencias
    FROM vw_emendas_temp
    GROUP BY `Código_da_Emenda`
    HAVING COUNT(*) > 1
    ORDER BY total_ocorrencias DESC
""").show(20)
```
